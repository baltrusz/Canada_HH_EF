# Import SHS 2017 Diary (F1) data

Loads `SHS-62M004X-E-2017-Diary_F1.csv` and takes a first look at its structure (shape, columns, dtypes, head).

```python
import os
import pandas as pd

DATA_DIR = r"C:\Users\marta.baltruszewicz\OneDrive - Asplan Viak\Documents\additional_projects\Canada_hh_energy\Data\OneDrive_2026-03-09\Survey of Household Spending\SHS_EDM_2017\Data - Données\CSV"
CSV_PATH = os.path.join(DATA_DIR, "SHS-62M004X-E-2017-Diary_F1.csv")

df = pd.read_csv(CSV_PATH)
print(f"shape: {df.shape}")
df.head()
```

```python
df.dtypes
```

```python
df.describe(include="all").T
```

## Weighted expenditure sums

Multiply every expenditure column (`CC001` through the last column, `TX001`) by `WeightD`, sum across households, and build a two-column table: expenditure code (column 1) and weighted sum (column 2). Export to xlsx.

```python
expenditure_cols = df.columns[df.columns.get_loc("CC001"):].tolist()

weighted_sums = df[expenditure_cols].mul(df["WeightD"], axis=0).sum()

weighted_expenditure = weighted_sums.reset_index()
weighted_expenditure.columns = ["expenditure_code", "weighted_sum"]

print(f"expenditure_cols: {len(expenditure_cols)} ({expenditure_cols[0]} .. {expenditure_cols[-1]})")
print(f"weighted_expenditure shape: {weighted_expenditure.shape}")
weighted_expenditure.head()
```

```python
OUT_PATH = os.path.join(DATA_DIR, "SHS_2017_Diary_weighted_expenditure.xlsx")
weighted_expenditure.to_excel(OUT_PATH, index=False)
print(f"Saved to: {OUT_PATH}")
```

# Import SHS 2021 data (SAS)
